```python
from sklearn.preprocessing import LabelEncoder

def force_numeric(df):
    df_fixed = df.copy()
    for col in df_fixed.columns:
        if not pd.api.types.is_numeric_dtype(df_fixed[col]):
            le = LabelEncoder()
            df_fixed[col] = le.fit_transform(df_fixed[col].astype(str))
    return df_fixed
```

```python
# News categories (arms per context)
news_categories = ["Entertainment", "Education", "Tech", "Crime"]

NUM_CONTEXTS = 3          # User1, User2, User3
ARMS_PER_CONTEXT = 4      # 4 news categories
TOTAL_ARMS = NUM_CONTEXTS * ARMS_PER_CONTEXT

print("Total arms:", TOTAL_ARMS)
```

```python
def get_arm_index(context_id, category_id):
    """
    context_id  : 0, 1, 2  (User1, User2, User3)
    category_id : 0, 1, 2, 3 (Entertainment, Education, Tech, Crime)
    """
    return context_id * ARMS_PER_CONTEXT + category_id
```

```python
from rlcmab_sampler import sampler

ROLL_NUMBER = 26   # ⚠️ replace with your actual roll number if different
reward_sampler = sampler(ROLL_NUMBER)
```

```python
import numpy as np

epsilon = 0.1

Q = np.zeros(TOTAL_ARMS)   # Estimated reward for each arm
N = np.zeros(TOTAL_ARMS)   # Number of times each arm is selected
```

```python
def run_epsilon_greedy(steps, epsilon):
    Q = np.zeros(TOTAL_ARMS)
    N = np.zeros(TOTAL_ARMS)
    rewards = []

    for t in range(steps):

        # Simulate a random context (0,1,2)
        context_id = np.random.randint(0, NUM_CONTEXTS)

        # Choose arm
        if np.random.rand() < epsilon:
            arm = np.random.randint(0, TOTAL_ARMS)
        else:
            arm = np.argmax(Q)

        # Get reward from sampler
        reward = reward_sampler.sample(arm)

        # Update counts
        N[arm] += 1
        Q[arm] += (reward - Q[arm]) / N[arm]

        rewards.append(reward)

    return np.array(rewards)
```

```python
def run_ucb(steps, c=2):
    Q = np.zeros(TOTAL_ARMS)
    N = np.zeros(TOTAL_ARMS)
    rewards = []

    for t in range(steps):

        context_id = np.random.randint(0, NUM_CONTEXTS)

        # Initialize each arm once
        if t < TOTAL_ARMS:
            arm = t
        else:
            confidence = c * np.sqrt(np.log(t + 1) / (N + 1e-5))
            arm = np.argmax(Q + confidence)

        reward = reward_sampler.sample(arm)

        N[arm] += 1
        Q[arm] += (reward - Q[arm]) / N[arm]

        rewards.append(reward)

    return np.array(rewards)
```

```python
def run_softmax(steps, tau=0.1):
    Q = np.zeros(TOTAL_ARMS)
    N = np.zeros(TOTAL_ARMS)
    rewards = []

    for t in range(steps):

        context_id = np.random.randint(0, NUM_CONTEXTS)

        # Compute softmax probabilities
        exp_Q = np.exp(Q / tau)
        probs = exp_Q / np.sum(exp_Q)

        arm = np.random.choice(np.arange(TOTAL_ARMS), p=probs)

        reward = reward_sampler.sample(arm)

        N[arm] += 1
        Q[arm] += (reward - Q[arm]) / N[arm]

        rewards.append(reward)

    return np.array(rewards)
```

```python
STEPS = 5000   # you can change this

eps_rewards = run_epsilon_greedy(STEPS, epsilon=0.1)
ucb_rewards = run_ucb(STEPS, c=2)
soft_rewards = run_softmax(STEPS, tau=0.1)
```

```python
def cumulative_average(rewards):
    return np.cumsum(rewards) / (np.arange(len(rewards)) + 1)

eps_avg = cumulative_average(eps_rewards)
ucb_avg = cumulative_average(ucb_rewards)
soft_avg = cumulative_average(soft_rewards)
```

```python
plt.figure(figsize=(10,6))

plt.plot(eps_avg, label="Epsilon-Greedy (ε=0.1)")
plt.plot(ucb_avg, label="UCB (c=2)")
plt.plot(soft_avg, label="SoftMax (τ=0.1)")

plt.xlabel("Steps")
plt.ylabel("Average Reward")
plt.title("Average Reward vs Time")
plt.legend()
plt.grid()
plt.show()
```

```python
eps_01 = cumulative_average(run_epsilon_greedy(STEPS, 0.1))
eps_02 = cumulative_average(run_epsilon_greedy(STEPS, 0.2))
eps_005 = cumulative_average(run_epsilon_greedy(STEPS, 0.05))

plt.figure(figsize=(10,6))
plt.plot(eps_01, label="ε=0.1")
plt.plot(eps_02, label="ε=0.2")
plt.plot(eps_005, label="ε=0.05")

plt.title("Epsilon Comparison")
plt.xlabel("Steps")
plt.ylabel("Average Reward")
plt.legend()
plt.show()
```
